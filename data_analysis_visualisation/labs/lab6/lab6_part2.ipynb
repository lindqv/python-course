```python
import pandas as pd

listings = pd.read_csv("../data/listings_clean.csv")
```

```python
mean = listings["review_scores_rating"].mean()
median = listings["review_scores_rating"].median()
std = listings["review_scores_rating"].std()

print("Mean", mean.round(2))
print("Median", median.round(2))
print("Std. dev.", std.round(2))
```

The mean is lower than the median due to the dataset having many low values.  
I would report the median since it gives a value that is more typical for this dataset.  
In terms of the value distribution, I would say that it has a couple very low values, and a lot of high values.

## Task 2

```python
q1 = listings["review_scores_rating"].quantile(0.25)
q3 = listings["review_scores_rating"].quantile(0.75)
iqr = q3 - q1

lower_fence = q1 - iqr * 1.5
print("Lower fence", lower_fence)
listings_below = listings[listings["review_scores_rating"] < lower_fence]

print("Listings below count", listings_below.shape[0])
```

```python
print(listings_below["review_scores_rating"].describe())
print(listings["review_scores_rating"].describe())
```

Median for listings below the lower fence: 4.00  
Median for all listings: 4.92

```python
five_reviews_or_fewer_count_below = listings_below[listings_below["number_of_reviews"] <= 5].shape[0]
count_below = listings_below.shape[0]
share_below = five_reviews_or_fewer_count_below / count_below

five_reviews_or_fewer_count = listings[listings["number_of_reviews"] <= 5].shape[0]
count = listings.shape[0]
share = five_reviews_or_fewer_count / count

listings_others = listings[listings["review_scores_rating"] >= lower_fence]
five_reviews_or_fewer_count_others = listings_others[listings_others["number_of_reviews"] <= 5].shape[0]
count_others = listings_others.shape[0]
share_others = five_reviews_or_fewer_count_others / count_others

print("Share with 5 reviews or lower, below the fence:", round(share_below, 2))
print("Share with 5 reviews or lower, all:", round(share, 2))
print("Share with 5 reviews or lower, others:", round(share_others, 2))
```

```python
listings_below.groupby("room_type")["number_of_reviews"].count()
```

I will keep the outliers but mark them with `True` in a boolean column called `rating_outlier`.

```python
listings["rating_outlier"] = listings["review_scores_rating"] < lower_fence
```

```python
print("Mean and median, unmodified:")
print(listings["review_scores_rating"].mean().round(2),
      listings["review_scores_rating"].median())

print("Mean and median if the lower outliers were to be removed:")
print(listings_others["review_scores_rating"].mean().round(2),
      listings_others["review_scores_rating"].median())
```
