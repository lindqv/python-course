```python
import pandas as pd

listings = pd.read_csv("../data/listings.csv")
```

```python
central = ["Södermalms", "Östermalms", "Norrmalms", "Kungsholmens"]
central_mask = listings["neighbourhood_cleansed"].isin(central)
private_room_mask_central = (central_mask) & (listings["room_type"] == "Private room")
private_room_mask_not_central = ~(central_mask) & (listings["room_type"] == "Private room")

central_listings = listings[central_mask]
central_count = len(central_listings)
total_listings = len(listings)
central_listings_share = central_count / total_listings

private_room_listings_central = listings[private_room_mask_central]
private_room_count_central = len(private_room_listings_central)
private_room_share_central = private_room_count_central / central_count

private_room_listings_not_central = listings[private_room_mask_not_central]
private_room_count_not_central = len(private_room_listings_not_central)
non_central_count = len(listings[~central_mask])
private_room_share_not_central = private_room_count_not_central / non_central_count

print("Central count", central_count)
print("Central listings share", central_listings_share)
print("Private room share central", private_room_share_central)
print("Private room share not central", private_room_share_not_central)
```
