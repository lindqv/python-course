# Task 1

```python
import pandas as pd

loans = pd.read_csv("../data/library_loans_clean.csv", parse_dates=["loan_date", "return_date"])
print(loans.shape)
print(loans.dtypes)

missing = loans.isna().sum()
print("Missing values, counts \n", missing)

missing_percentages = loans.isna().mean() * 100
print("Missing values, percentages \n", missing_percentages)
```

# Task 2

```python
loan_date_missing_mask = loans["loan_date"].isna()
loan_date_missing = loans[loan_date_missing_mask]

return_date_missing_returned = loans[loans["return_date"].isna() & loans["returned"]]
return_date_missing_not_returned = loans[loans["return_date"].isna() & ~loans["returned"]]

print(loan_date_missing)
print("Missing return date and returned count", len(return_date_missing_returned))
print("Missing return date and not returned count", len(return_date_missing_not_returned))
```

```python
loans["has_return_date"] = loans["return_date"].isna()
```

```python
loans["has_return_date"].value_counts()
```

# Task 3

```python
members_raw = pd.read_csv("../data/library_members.csv")
members = members_raw.copy()
print(members.shape)
print(members["birth_year"].info())
print(members["birth_year"].describe())

members_min = members[members["birth_year"] == 1900]
members_max = members[members["birth_year"] == 2030]
members_min, members_max

print("Number of member with min value", len(members_min))
print("Number of member with max value", len(members_max))
```

```python
duplicate_rows = members[members.duplicated()]
duplicate_member_id = members[members["member_id"].duplicated()]

print(duplicate_rows)
print(duplicate_member_id)

print("Duplicate count members", len(duplicate_rows))
print("Duplicate count member id", len(duplicate_member_id))
```

The minimum and maximum birth year values are 1900 (count 8) and 2030 (count 3), values that would not be applicable to any currently living member.  

There are 4 duplicates in the members table, and 4 in the member id column.

# Task 4

```python
members["birth_year"] = members["birth_year"].where(members["birth_year"].between(1920, 2015))

print("Missing values now", members["birth_year"].isna().sum())
members.describe()
```

# Task 5

a)

```python
print(members.dropna())
print(members.dropna(subset=["branch"]))
print(members.dropna(subset=["birth_year"]))
```

b)

```python
members["branch"] = members["branch"].fillna("Unknown")
members["branch"].value_counts()
```

c)

```python
print("Mean before", members["birth_year"].mean().round(2))
print("Standard deviation before", members["birth_year"].std().round(2))

median_year = members["birth_year"].median()
median_filled = members["birth_year"].fillna(median_year)
print("Median year", median_year)

print("Mean after", median_filled.mean().round(2))
print("Standard deviation after", median_filled.std().round(2))
```

# Task 6

```python
print("Loans shape", loans.shape)
print("Members shape", members.shape)
```

```python
first_merge = loans.merge(members, how="left")
print(first_merge.shape)
```

```python
late_fee_total_before = loans["late_fee"].sum()
late_fee_total_after = first_merge["late_fee"].sum()
print("Late fee total before and after", late_fee_total_before, late_fee_total_after)
print("Difference", abs(late_fee_total_before - late_fee_total_after))
```

```python
second_merge = loans.merge(members, how="left", validate="many_to_one")
```

# Task 7

```python
members_no_duplicates = members.drop_duplicates()
print(members_no_duplicates.shape)

third_merge = loans.merge(members_no_duplicates, how="left", validate="many_to_one", indicator=True)
print(third_merge.shape)
print("Late fee total", third_merge["late_fee"].sum())

print(third_merge["_merge"].value_counts())
left_only_members = third_merge[third_merge["_merge"] == "left_only"] 
print("Number of left only members", len(left_only_members))
print("Unique members with left only", left_only_members["member_id"].unique())
print("Late fees from left only", left_only_members["late_fee"].sum())
```

```python
inner_merge = loans.merge(members_no_duplicates, how="inner", validate="many_to_one")
print(inner_merge.shape)
print("Late fee total", inner_merge["late_fee"].sum())
```
