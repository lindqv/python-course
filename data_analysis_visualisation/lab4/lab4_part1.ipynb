# Task 1

```python
import pandas as pd

loans = pd.read_csv("../data/library_loans_clean.csv", parse_dates=["loan_date", "return_date"])
print(loans.shape)
print(loans.dtypes)

missing = loans.isna().sum()
print("Missing values, counts \n", missing)

missing_percentages = loans.isna().mean() * 100
print("Missing values, percentages \n", missing_percentages)
```

# Task 2

```python
loan_date_missing_mask = loans["loan_date"].isna()
loan_date_missing = loans[loan_date_missing_mask]

return_date_missing_returned = loans[loans["return_date"].isna() & loans["returned"]]
return_date_missing_not_returned = loans[loans["return_date"].isna() & ~loans["returned"]]

print(loan_date_missing)
print("Missing return date and returned count", len(return_date_missing_returned))
print("Missing return date and not returned count", len(return_date_missing_not_returned))
```
