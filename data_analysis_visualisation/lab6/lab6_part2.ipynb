```python
import pandas as pd

listings = pd.read_csv("../data/listings_clean.csv")
```

```python
mean = listings["review_scores_rating"].mean()
median = listings["review_scores_rating"].median()
std = listings["review_scores_rating"].std()

print("Mean", mean.round(2))
print("Median", median.round(2))
print("Std. dev.", std.round(2))
```

The mean is lower than the median due to the dataset having many low values.  
I would report the median since it gives a value that is more typical for this dataset.  
In terms of the value distribution, I would say that it has a couple very low values, and a lot of high values.
