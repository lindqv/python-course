```python
import pandas as pd

listings = pd.read_csv("../data/listings.csv")
```

# Question 1

```python
listings["has_reviews"] = listings["number_of_reviews"] > 0
listings[["has_reviews", "number_of_reviews"]].head()
counts = listings["has_reviews"].value_counts()
shares = listings["has_reviews"].value_counts(normalize=True).round(3)

print(counts, shares)
```

Share of listings without reviews: 14.1 %

```python
listings[["neighbourhood_cleansed", "has_reviews", "number_of_reviews"]].head()
shares = listings.groupby("neighbourhood_cleansed")["has_reviews"].value_counts(normalize=True).round(3)
shares
```

Neighbourhood with lowest share without review: Södermalm, 0.076  
Neighbourhood with highest share without review: Hässelby-Vällingby, 0.270
