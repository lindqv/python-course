# Question 3

```python
listings.groupby("room_type")["beds"].mean().round(2)
listings.groupby("room_type")["beds"].agg(["count", "mean"]).round(2)
```

The mean for hotel room and shared room only have 3 and 24 datapoints, respectively. So we cannot be sure that the mean is representative for these two room types as a whole, since there are few entries for them. The average I trust the least is the one for hotel rooms with 3 entry counts.

## Question 4

```python
listings.groupby("neighbourhood_cleansed")["number_of_reviews"].agg(
    ["count", "mean", "median"]
    ).round(2).sort_values("median", ascending=False)
```

Highest median: Södermalms  
Lowest median: Skärholmens  
Södermalms having a larger mean than median is probably caused by a few listings having a much larger mean than the others, which raises the mean but does not affect the median.
